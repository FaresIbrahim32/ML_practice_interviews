```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
torch.manual_seed(42)
X = torch.rand(100, 1) * 10
y = 2 * X + 3 + torch.randn(100, 1)
```

```python
X.shape,y.shape
```

```python
class MyLoss(nn.Module):
    def __init__(self, th):
        super().__init__()
        self.th = th

    def forward(self, y_pred, y_true):
        pen = torch.abs(y_true - y_pred)


        quadratic = torch.square(y_true - y_pred) * 0.5
        linear = self.th * (torch.abs(y_true - y_pred) - (0.5 * self.th))


        rem = torch.where(pen <= self.th, quadratic, linear)

        return rem.mean()
```

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)
```

```python
model = LinearRegressionModel()
huber_loss = MyLoss(th=1.0)
criterion = huber_loss
optimizer = optim.SGD(model.parameters(), lr=0.01)


epochs = 1000
for epoch in range(epochs):

    predictions = model(X)
    loss = criterion(predictions, y)


    optimizer.zero_grad()
    loss.backward()
    optimizer.step()


    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
[w, b] = model.linear.parameters()
print(f"Learned weight: {w.item():.4f}, Learned bias: {b.item():.4f}")


X_test = torch.tensor([[4.0], [7.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```
